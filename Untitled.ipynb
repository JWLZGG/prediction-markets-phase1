```python
def classify_terminal_prices(prices, eps=1e-3):
    try:
        vals = [float(p) for p in prices]
    except Exception:
        return "invalid"

    if len(vals) != 2:
        return "invalid"

    if abs(vals[0] - 1.0) < eps and abs(vals[1] - 0.0) < eps:
        return "yes_won"
    if abs(vals[0] - 0.0) < eps and abs(vals[1] - 1.0) < eps:
        return "no_won"
    if vals[0] == 0.0 and vals[1] == 0.0:
        return "double_zero"
    return "ambiguous"

tmp["label_status"] = tmp["parsed_outcome_prices"].apply(classify_terminal_prices)
tmp["label_status"].value_counts(dropna=False)
```

```python
def infer_confident_label(prices, eps=1e-3):
    try:
        vals = [float(p) for p in prices]
    except Exception:
        return None

    if len(vals) != 2:
        return None

    if abs(vals[0] - 1.0) < eps and abs(vals[1] - 0.0) < eps:
        return 1   # Yes won
    if abs(vals[0] - 0.0) < eps and abs(vals[1] - 1.0) < eps:
        return 0   # No won
    return None

tmp["resolved_outcome_confident"] = tmp["parsed_outcome_prices"].apply(infer_confident_label)

tmp[["market_id", "question", "parsed_outcomes", "parsed_outcome_prices", "resolved_outcome_confident"]].head(20)
```
